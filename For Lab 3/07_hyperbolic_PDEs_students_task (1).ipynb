## Problem Assignment: Numerical Schemes for the 1D Linear Advection Equation

### 1\. Problem Definition

You are tasked with numerically solving the 1D linear advection equation:

$$
\frac{\partial u}{\partial t} + \frac{\partial u}{\partial x} = 0
$$

This equation describes a wave $u(x, t)$ traveling to the right with a constant speed $c=1$.

* **Domain:** A periodic spatial domain, $x \in [0, 25.0]$.
* **Initial Condition ($t=0$):** The solution starts as a sum of two pulses:
$u(x, 0) = \exp(-20.0(x - 2)^2) + \exp(-(x - 5)^2)$
* **Exact Solution:** For this problem, the exact solution is the initial shape simply "advecting" or moving to the right. It is given by $u(x, t) = u(x-t, 0)$.

We will use the following Python function to represent the exact solution at any given time `t`:

```python
import numpy as np
u_true = lambda x, t: np.exp(-20.0 * ((x - t) - 2.0)**2) + np.exp(-((x - t) - 5.0)**2)
```

````python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Problem Definition ---
u_true = lambda x, t: np.exp(-20.0 * ((x - t) - 2.0)**2) + np.exp(-((x - t) - 5.0)**2)

# --- 2. Numerical Setup ---
m = 501
x = np.linspace(0, 25.0, m)
delta_x = 25.0 / (m - 1)
cfl = 0.8 # This is our nu
delta_t = cfl * delta_x
t = 0.0
t_final = 17.0

# --- 3. Run Simulations ---

# Method 1: FE
U_fe = u_true(x, 0)
U_new = np.empty(U_fe.shape)
t = 0.0
while t < t_final:
    # ... IMPLEMENT FTCS loop here ...
    U_new[1:-1] = (U_fe[1:-1]- (cfl / 2.0) * (U_fe[2:] - U_fe[:-2]))
    
    # ... Remember periodic boundaries ...
    U_new[0] = U_fe[0] - (cfl / 2.0) * (U_fe[1] - U_fe[-1])
    U_new[-1] = U_fe[-1] - (cfl / 2.0) * (U_fe[0] - U_fe[-2])
    
    U_fe = U_new.copy()
    t += delta_t

# Method 2: Lax-Friedrichs
U_lax_friedrichs = u_true(x, 0)
U_new = np.empty(U_lax_friedrichs.shape)
t = 0.0
while t < t_final:
    # ... IMPLEMENT Lax-Friedrichs loop here ...
    # This is the one from your example
    U_new[1:-1] = 0.5 * (U_lax_friedrichs[2:] + U_lax_friedrichs[:-2]) - (cfl / 2.0) * (U_lax_friedrichs[2:] - U_lax_friedrichs[:-2])
    U_new[0] = 0.5 * (U_lax_friedrichs[1] + U_lax_friedrichs[-1]) - (cfl / 2.0) * (U_lax_friedrichs[1] - U_lax_friedrichs[-1])
    U_new[-1] = 0.5 * (U_lax_friedrichs[0] + U_lax_friedrichs[-2]) - (cfl / 2.0) * (U_lax_friedrichs[0] - U_lax_friedrichs[-2])
    
    U_lax_friedrichs = U_new.copy()
    t += delta_t

# Method 3: Upwind
U_upwind = u_true(x, 0)
U_new = np.empty(U_upwind.shape)
t = 0.0
while t < t_final:
    # ... IMPLEMENT Upwind loop here ...
    U_new[1:] = (U_upwind[1:] - cfl * (U_upwind[1:] - U_upwind[:-1]))

    # Periodic boundary condition
    U_new[0] = U_upwind[0] - cfl * (U_upwind[0] - U_upwind[-1])

    U_upwind = U_new.copy()
    t += delta_t

# Method 4: Lax-Wendroff
U_lax_wendroff = u_true(x, 0)
U_new = np.empty(U_lax_wendroff.shape)
t = 0.0
while t < t_final:
    # ... IMPLEMENT Lax-Wendroff loop here ...
    U_new[1:-1] = (U_lax_wendroff[1:-1]- (cfl / 2.0) * (U_lax_wendroff[2:] - U_lax_wendroff[:-2])+ cfl**2 / 2.0 * (U_lax_wendroff[2:]- 2.0 * U_lax_wendroff[1:-1]+ U_lax_wendroff[:-2]))
    
    # Periodic boundary conditions
    U_new[0] = (U_lax_wendroff[0]- cfl / 2.0 * (U_lax_wendroff[1] - U_lax_wendroff[-1])+ cfl**2 / 2.0 * (U_lax_wendroff[1]- 2.0 * U_lax_wendroff[0]+ U_lax_wendroff[-1]))
    U_new[-1] = (U_lax_wendroff[-1]- cfl / 2.0 * (U_lax_wendroff[0] - U_lax_wendroff[-2])+ cfl**2 / 2.0 * (U_lax_wendroff[0]- 2.0 * U_lax_wendroff[-1]+ U_lax_wendroff[-2]))
    
    U_lax_wendroff = U_new.copy()
    t += delta_t

# Method 5: Beam-Warming
U_beam_warming = u_true(x, 0)
U_new = np.empty(U_beam_warming.shape)
t = 0.0
while t < t_final:
    # ... IMPLEMENT Beam-Warming loop here ...
    U_new[2:] = (U_beam_warming[2:]- cfl / 2.0 * (3.0 * U_beam_warming[2:]- 4.0 * U_beam_warming[1:-1]+ U_beam_warming[:-2])+ cfl**2 / 2.0 * (U_beam_warming[2:]- 2.0 * U_beam_warming[1:-1]+ U_beam_warming[:-2]))

    # Periodic boundary conditions: i = 0
    U_new[0] = (U_beam_warming[0] - cfl / 2.0 * (3.0 * U_beam_warming[0]- 4.0 * U_beam_warming[-1]+ U_beam_warming[-2])+ cfl**2 / 2.0 * (U_beam_warming[0]- 2.0 * U_beam_warming[-1]+ U_beam_warming[-2]))
    # Periodic boundary conditions: i = 1
    U_new[1] = (U_beam_warming[1]- cfl / 2.0 * (3.0 * U_beam_warming[1]- 4.0 * U_beam_warming[0]+ U_beam_warming[-1])+ cfl**2 / 2.0 * (U_beam_warming[1]- 2.0 * U_beam_warming[0]+ U_beam_warming[-1]))
    
    U_beam_warming = U_new.copy()
    t += delta_t

# --- 4. Plotting ---
fig = plt.figure(figsize=(12, 8))
axes = fig.add_subplot(1, 1, 1)

# Plot the exact solution
axes.plot(x, u_true(x, t_final), 'k', label='Exact Solution')

# Plot your numerical solutions
# ... PLOT U_ftcs ...
axes.plot(x, U_fe, 'm--', label='FTCS')
# ... PLOT U_lax_friedrichs ...
axes.plot(x, U_lax_friedrichs, 'ro', label='Lax-Friedrichs', markersize=4)
# ... PLOT U_upwind ...
axes.plot(x, U_upwind, 'b--', label='Upwind')
# ... PLOT U_lax_wendroff ...
axes.plot(x, U_lax_wendroff, 'g-', label='Lax-Wendroff')
# ... PLOT U_beam_warming ...
axes.plot(x, U_beam_warming, 'c-.', label='Beam-Warming')

# Set limits and labels
axes.set_xlim((15.0, 25.0))
axes.set_ylim((-0.1, 1.1))
axes.set_xlabel('x')
axes.set_ylabel('u(x, t)')
axes.set_title(f'Numerical Methods Comparison at t = {t_final:.1f}')
axes.legend()

plt.show()
````
